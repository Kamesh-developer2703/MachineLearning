```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df=pd.read_csv("dataset.csv")
df.head(4)
```

```python
x=df.drop('price',axis=1)
y=df['price']
```

```python
x.shape
```

```python
x.head(10)
```

```python
y.tail(3)
```

```python
x.shape
```

```python
y.shape
```

```python
model = LinearRegression()
model.fit(x,y)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
%matplotlib inline
```

```python
plt.xlabel("Area")
plt.ylabel("Price")
plt.scatter(x,y)
```

## Prediction

```python
area = int(input("Enter the area of the house: "))
area=pd.DataFrame([[area]],columns=["area"])
```

```python
model.predict(area)
```

# Regression Line Y

```python
y=model.coef_*area+model.intercept_
```

```python
y
```

```python
print("The Price of {0} Square Feet Land is {1}".format(area,y))
```

## Prediction Graph

```python
plt.xlabel("Area of House")
plt.ylabel("Predicted Price")
plt.scatter(area,y)
```
