```
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score
```

```
dataset = load_iris()
```

# display the dataset

```
print(dataset.data)
```

```
print(dataset.target)
```

```
print(dataset.feature_names)
```

```
x=pd.DataFrame(dataset.data,columns=dataset.feature_names)
print(x.head())
```

```
column_name = ["flower"]
y=pd.DataFrame(dataset.target,columns=column_name)
y
```

```
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.75,random_state=0)
```

```
print(x_train.head())
```

```
print(x_train.shape)
```

```
print(x_test.shape)
```

# Find best max_length

```
accuracy=[]
for i in range(1,20):
    model=DecisionTreeClassifier(max_depth=i,random_state=0)
    model.fit(x_train,y_train)
    model.fit(x_train,y_train)
    y_pred = model.predict(x_test)
    score = accuracy_score(y_test,y_pred)
    accuracy.append(score)
    print(accuracy)
best_depth = accuracy.index(max(accuracy)) + 1
best_depth
```

```
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(range(1, 20), accuracy, marker='o')
plt.xlabel("Max Depth")
plt.ylabel("Accuracy")
plt.title("Accuracy vs Max Depth")
plt.show()
```

```
model = DecisionTreeClassifier(max_depth=best_depth,random_state=0)
model.fit(x_train,y_train)
```

```
y_pred = model.predict(x_test)
y_pred = pd.DataFrame(y_pred,columns=['My Result'])
```

```
y_pred
```

```
print(accuracy_score(y_test,y_pred)*100)
```

```
import math as m
print("Accuracy Score : ",m.ceil(accuracy_score(y_test,y_pred)*100),"%")
```
